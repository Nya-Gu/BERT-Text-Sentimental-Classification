```python
# import torch.nn as nn

# # Temperature 스케일링 값 탐색 (valid_klue_logits, valid_kykim_logits, valid_roberta_logits, valid_beomi_logits, valid_koelectra_logits)
# temperature = nn.Parameter(torch.ones(1, device='cuda') * 1.5)
# optimizer = torch.optim.LBFGS([temperature], lr=0.01, max_iter=50)

# def eval():
#     optimizer.zero_grad()
#     loss = nn.CrossEntropyLoss()(valid_koelectra_logits / temperature, valid_targets)
#     loss.backward()
#     return loss

# optimizer.step(eval)
# print(f"최적의 T: {temperature.item():.4f}")
```

```python
klue_T      = 1.0995
kykim_T     = 1.2114
roberta_T   = 1.1614
beomi_T     = 1.1812
koelectra_T = 1.1846

prob_klue       = torch.softmax(valid_klue_logits / klue_T,             dim=-1)
prob_kykim      = torch.softmax(valid_kykim_logits / kykim_T,           dim=-1)
prob_roberta    = torch.softmax(valid_roberta_logits / roberta_T,       dim=-1)
prob_beomi      = torch.softmax(valid_beomi_logits / beomi_T,           dim=-1)
prob_koelectra  = torch.softmax(valid_koelectra_logits / koelectra_T,   dim=-1)

probs_list = [prob_klue, prob_kykim, prob_roberta, prob_beomi, prob_koelectra]
all_probs = torch.stack(probs_list, dim=1).cuda()

# Ablation study용
off_klue       = 0
off_kykim      = 0
off_roberta    = 0
off_beomi      = 0
off_koelectra  = 0

best_acc = 0
best_weight = None

for i in range(11):
    print(f"{i}/10")
    if off_klue and i != 0:
        continue

    for j in range(11):
        if off_kykim and j != 0:
            continue

        for k in range(11):
            if off_roberta and k != 0:
                continue

            for l in range(11):
                if off_beomi and l != 0:
                    continue

                for m in range(11):
                    if off_koelectra and m != 0:
                        continue

                    model_weight   = 0.1 * torch.tensor([i,j,k,l,m]).view(1,5,1).cuda()
                    weighted_probs = torch.sum(model_weight * all_probs, dim=1)
                    weighted_probs = weighted_probs.argmax(dim=-1)
                    soft_acc       = (weighted_probs == valid_targets).float().mean().item() * 100

                    if soft_acc > best_acc:
                        best_acc = soft_acc
                        best_weight = model_weight

print(f"soft acc: {best_acc:.4f}")
print(f"best weight: {best_weight}")
```

## 6-3. Hard Voting

```python
import torch.nn.functional as F

pred_klue       = F.one_hot(valid_klue_logits.argmax(dim=-1),      num_classes=4)
pred_kykim      = F.one_hot(valid_kykim_logits.argmax(dim=-1),     num_classes=4)
pred_roberta    = F.one_hot(valid_roberta_logits.argmax(dim=-1),   num_classes=4)
pred_beomi      = F.one_hot(valid_beomi_logits.argmax(dim=-1),     num_classes=4)
pred_koelectra  = F.one_hot(valid_koelectra_logits.argmax(dim=-1), num_classes=4)

hard_pred = pred_klue + pred_kykim + pred_roberta + pred_beomi + pred_koelectra
hard_pred = hard_pred.argmax(dim=-1)

hard_acc = (hard_pred == valid_targets).float().mean().item() * 100
print(f"hard acc: {hard_acc:.4f}")
```

## 6-4. Stacking

```python
import torch.nn as nn

# Temperature 스케일링 값 탐색
temperature = nn.Parameter(torch.ones(1, device='cuda') * 1.5)
optimizer = torch.optim.LBFGS([temperature], lr=0.01, max_iter=50)

def eval():
    optimizer.zero_grad()
    loss = nn.CrossEntropyLoss()(valid_koelectra_logits / temperature, valid_targets)
    loss.backward()
    return loss

optimizer.step(eval)
print(f"최적의 T: {temperature.item():.4f}")
```
